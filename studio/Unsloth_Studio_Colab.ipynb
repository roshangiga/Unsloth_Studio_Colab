```python
import sys
sys.path.insert(0, "/content/unsloth/studio/backend")
from colab import start
start()
```

```python
import os
import shutil

output_dir = "/root/.unsloth/studio/outputs/"
target_dir = "./models"

# Create the target directory if it doesn't exist
os.makedirs(target_dir, exist_ok=True)

# Copy all contents from the output directory to the target directory
for item in os.listdir(output_dir):
    s = os.path.join(output_dir, item)
    d = os.path.join(target_dir, item)
    if os.path.isdir(s):
        shutil.copytree(s, d, dirs_exist_ok=True)
    else:
        shutil.copy2(s, d)

print(f"Models copied from {output_dir} to {target_dir}")
```
